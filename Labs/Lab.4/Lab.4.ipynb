1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class Counter:
    def __init__(self, max_value:int):
        self.max_value = max_value
        self.count = 0
    def add_count(self) -> bool:
        if self.count > self.max_value:
            print(f"Error: Cannot add more! At maximum!")
            return False
        self.count += 1
        return True
    def reset_count(self):
        self.count = 0
        return f"Counter is reset to {self.count}"
    
    
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the perimeter of the object. 

```python
from abc import ABC, abstractmethod
import math
class Base(ABC):
    @abstractmethod
    def get_area(self):
        pass
    @abstractmethod
    def get_perimeter(self):
        pass
    @abstractmethod
    def generate_points(self, num_points = 16):
        pass

class Circle(Base):
    def __init__(self, radius:float, x:float=0.0, y:float=0.0):
        self.__radius = radius
        self.__x = x
        self.__y = y

    def get_area(self):
        return math.pi*(self.__radius**2)

    def get_perimeter(self):
        return 2*math.pi * self.__radius

    def get_radius(self):
        return self.__radius

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def generate_points(self, num_points=16):
        points = []
        for i in range(num_points):
            angle = 2*math.pi * i/num_points
            pt_x = self.__x + self.__radius * math.cos(angle)
            pt_y = self.__y + self.__radius * math.sin(angle)
            points.appednd((round(pt_x, 2), round(pt_y,2)))
        return points

class Rectangle(Base):
    def __init__(self, length:float, width:float, x:float = 0, y:float = 0):
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    def get_area(self):
        area = (self.__length * self.__width)
        return f"Area of the rectangle is {area} units sqaured"

    def get_perimeter(self):
        perimeter = (2*self.__length) + (2*self.__width)
        return f"Perimeter of the rectangle is {perimeter} units"

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def generate_points(self, num_points=16):
        points = []
        perimeter = self.get_perimeter()
        step = perimeter/num_points

        for i in range(num_points):
            distance = i * step
            if distance <= self.__width:
                pt_x = self.__x + distance
                pt_y = self.__y +self.__length
            elif distance <= self.__width + self.__length:
                pt_x = self.__x + distance
                pt_y = self.__y + self.__length - (distance - self.__width)
            elif distance <= 2* self.__width + self.__length:
                pt_x = self.__x + self.__width - (distance - (self.__width + self.__length))
                pt_y = self.__y
            else:
                pt_x = self.__x
                pt_y = self.__y + (distance - (2*self.__width + self.__length))
            points.append((round(pt_x, 2), round(pt_y,2)))
        return points
            

class Triangle(Base):
    def __init__(self, x1, y1, x2, y2, x3, y3):
        self.__x1, self.__y1 = x1 , y1
        self.__x2, self.__y2 = x2 , y2
        self.__x3, self.__y3 = x3 , y3

    def get_vertices(self):
        return [(self.__x1, self.__y1), (self.__x2, self.__y2), (self.__x3, self.__y3)]

    def __side_length(self, x_a, y_a, x_b, y_b):
        return math.hypot(x_b - x_a, y_b - y_a)
    
    def get_perimeter(self):
        side1 = self.__side_length(self.__x1, self.__y1, self.__x2, self.__y2)
        side2 = self.__side_length(self.__x2, self.__y2, self.__x3, self.__y3)
        side3 = self.__side_length(self.__x3, self.__y3, self.__x1, self.__y1)
        perimeter = side1 + side2 + side3
        return perimeter
    
    def get_area(self):
        area = 0.5 * abs(
            self.__x1 * (self.__y2 - self.__y3) + 
            self.__x2 * (self.__y3 - self.__y1) + 
            self.__x3 * (self.__y1 - self.__y2)
        )
        return area
    
    def generate_points(self, num_points=16):
        points = []
        v = [(self.__x1, self.__y1), (self.__x2, self.__y2), (self.__x3, self.__y3)]
        side1 = self.__side_length(v[0][0], v[0][1], v[1][0], v[1][1])
        side2 = self.__side_length(v[1][0], v[1][1], v[2][0], v[2][1])
        side3 = self.__side_length(v[2][0], v[2][1], v[0][0], v[0][1])
        
        perimeter = side1 + side2 + side3
        step = perimeter / num_points

        for i in range(num_points):
            distance = i * step
            if distance <= side1:
                t = distance/side1 if side1 > 0 else 0
                pt_x = v[0][0] + t * (v[1][0] - v[0][0])
                pt_y = v[0][1] + t * (v[1][1] - v[0][1])
            elif distance <= side1 + side2:
                t = (distance - side1)/side2 if side2 > 0 else 0
                pt_x = v[1][0] + t * (v[2][0] - v[1][0])
                pt_y = v[1][1] + t * (v[2][1] - v[1][1])
            else:
                t = (distance - side1 - side2)/side3 if side3 > 0 else 0
                pt_x = v[2][0] + t * (v[0][0] - v[2][0])
                pt_y = v[2][1] + t * (v[0][1] - v[2][1])
            points.append((round(pt_x, 2), round(pt_y,2)))
        return points

    
```

8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
from abc import ABC, abstractmethod
import math
class Base(ABC):
    @abstractmethod
    def get_area(self):
        pass
    @abstractmethod
    def get_perimeter(self):
        pass
    @abstractmethod
    def generate_points(self, num_points = 16):
        pass
    @abstractmethod
    def point_in_shape(self, px:float, py:float) -> bool:
        pass
class Circle(Base):
    def __init__(self, radius:float, x:float=0.0, y:float=0.0):
        self.__radius = radius
        self.__x = x
        self.__y = y

    def get_area(self):
        return math.pi*(self.__radius**2)

    def get_perimeter(self):
        return 2*math.pi * self.__radius

    def get_radius(self):
        return self.__radius

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def generate_points(self, num_points=16):
        points = []
        for i in range(num_points):
            angle = 2*math.pi * i/num_points
            pt_x = self.__x + self.__radius * math.cos(angle)
            pt_y = self.__y + self.__radius * math.sin(angle)
            points.appednd((round(pt_x, 2), round(pt_y,2)))
        return points

    def point_in_shape(self, px:float, py:float) -> bool:
        distance = math.hypot(px - self.__x, py-self.__y)
        return distance <= self.__radius

class Rectangle(Base):
    def __init__(self, length:float, width:float, x:float = 0, y:float = 0):
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y

    def get_area(self):
        area = (self.__length * self.__width)
        return f"Area of the rectangle is {area} units sqaured"

    def get_perimeter(self):
        perimeter = (2*self.__length) + (2*self.__width)
        return f"Perimeter of the rectangle is {perimeter} units"

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def generate_points(self, num_points=16):
        points = []
        perimeter = self.get_perimeter()
        step = perimeter/num_points

        for i in range(num_points):
            distance = i * step
            if distance <= self.__width:
                pt_x = self.__x + distance
                pt_y = self.__y +self.__length
            elif distance <= self.__width + self.__length:
                pt_x = self.__x + distance
                pt_y = self.__y + self.__length - (distance - self.__width)
            elif distance <= 2* self.__width + self.__length:
                pt_x = self.__x + self.__width - (distance - (self.__width + self.__length))
                pt_y = self.__y
            else:
                pt_x = self.__x
                pt_y = self.__y + (distance - (2*self.__width + self.__length))
            points.append((round(pt_x, 2), round(pt_y,2)))
        return points

    def point_in_shape(self, px:float,py:float) -> bool:
        in_x = self.__x <= px <= (self.__x + self.__width)
        in_y = self.__y <= py <= (self.__y + self.__length)
        return in_x and in_y

class Triangle(Base):
    def __init__(self, x1, y1, x2, y2, x3, y3):
        self.__x1, self.__y1 = x1 , y1
        self.__x2, self.__y2 = x2 , y2
        self.__x3, self.__y3 = x3 , y3

    def get_vertices(self):
        return [(self.__x1, self.__y1), (self.__x2, self.__y2), (self.__x3, self.__y3)]

    def __side_length(self, x_a, y_a, x_b, y_b):
        return math.hypot(x_b - x_a, y_b - y_a)
    
    def get_perimeter(self):
        side1 = self.__side_length(self.__x1, self.__y1, self.__x2, self.__y2)
        side2 = self.__side_length(self.__x2, self.__y2, self.__x3, self.__y3)
        side3 = self.__side_length(self.__x3, self.__y3, self.__x1, self.__y1)
        perimeter = side1 + side2 + side3
        return perimeter
    
    def get_area(self):
        area = 0.5 * abs(
            self.__x1 * (self.__y2 - self.__y3) + 
            self.__x2 * (self.__y3 - self.__y1) + 
            self.__x3 * (self.__y1 - self.__y2)
        )
        return area
    
    def generate_points(self, num_points=16):
        points = []
        v = [(self.__x1, self.__y1), (self.__x2, self.__y2), (self.__x3, self.__y3)]
        side1 = self.__side_length(v[0][0], v[0][1], v[1][0], v[1][1])
        side2 = self.__side_length(v[1][0], v[1][1], v[2][0], v[2][1])
        side3 = self.__side_length(v[2][0], v[2][1], v[0][0], v[0][1])
        
        perimeter = side1 + side2 + side3
        step = perimeter / num_points

        for i in range(num_points):
            distance = i * step
            if distance <= side1:
                t = distance/side1 if side1 > 0 else 0
                pt_x = v[0][0] + t * (v[1][0] - v[0][0])
                pt_y = v[0][1] + t * (v[1][1] - v[0][1])
            elif distance <= side1 + side2:
                t = (distance - side1)/side2 if side2 > 0 else 0
                pt_x = v[1][0] + t * (v[2][0] - v[1][0])
                pt_y = v[1][1] + t * (v[2][1] - v[1][1])
            else:
                t = (distance - side1 - side2)/side3 if side3 > 0 else 0
                pt_x = v[2][0] + t * (v[0][0] - v[2][0])
                pt_y = v[2][1] + t * (v[0][1] - v[2][1])
            points.append((round(pt_x, 2), round(pt_y,2)))
        return points

    def point_in_shape(self, px:float, py:float) -> bool:
        def sign(p1_x, p1_y, p2_x, p2_y, p3_x, p3_y):
            return (p1_x - p3_x) *(p2_y - p3_y) - (p2_x - p3_x) * (p1_y - p3_y)
        d1 = sign(px, py, self.__x1, self.__y1, self.__x2, self.__y2)
        d2 = sign(px, py, self.__x2, self.__y2, self.__x3, self.__y3)
        d3 = sign(px, py, self.__x3, self.__y3, self.__x1, self.__y1)
        has_neg = (d1 < 0) or (d2 < 0) or (d3 < 0)
        has_pos = (d1 > 0) or (d2 > 0) or (d3 > 0)
        return not(has_neg and has_pos)
```

```python
from abc import ABC, abstractmethod
import math
class Base(ABC):
    @abstractmethod
    def get_area(self):
        pass
        
    @abstractmethod
    def point_in_shape(self, px:float, py:float) -> bool:
        pass
        
    @abstractmethod
    def draw(self, canvas:"Canvas", char: str="*") -> None:
        pass

    @abstractmethod
    def __repr__(self) -> str:
        pass

        
class Circle(Base):
    def __init__(self, radius:float, x:float=0.0, y:float=0.0, symbol: str = "O"):
        self.__radius = radius
        self.__x = x
        self.__y = y
        self.symbol = symbol

    def get_area(self):
        return math.pi*(self.__radius**2)

    def get_radius(self):
        return self.__radius

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def point_in_shape(self, px:float, py:float) -> bool:
        distance = math.hypot(px - self.__x, py-self.__y)
        return distance <= self.__radius

    def draw(self, canvas:"Canvas", char: str = "*") -> None:
        for row in range(canvas.height):
            for col in range(canvas.width):
                if self.point_in_shape(col, row):
                    canvas.set_pixel(row, col, char)

    def __repr__(self) -> str:
        return(f"Circle(radius={self.__radius}, x = {self.__x}, y={self.__y}, symbol = {self.symbol!r})")

class Rectangle(Base):
    def __init__(self, length:float, width:float, x:float = 0, y:float = 0, symbol: str = "#"):
        self.__length = length
        self.__width = width
        self.__x = x
        self.__y = y
        self.symbol = symbol

    def get_area(self):
        area = (self.__length * self.__width)
        return f"Area of the rectangle is {area} units sqaured"

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def point_in_shape(self, px:float,py:float) -> bool:
        in_x = self.__x <= px <= (self.__x + self.__width)
        in_y = self.__y <= py <= (self.__y + self.__length)
        return in_x and in_y

    def draw(self, canvas: "Canvas", char: str="*") -> None:
        for row in range(canvas.height):
            for col in range(canvas.width):
                if self.point_in_shape(col,row):
                    canvas.set_pixel(row,col,char)
    
    def __repr__(self) -> str:
        return( f"Rectangle(width={self.__width}, length={self.length}, x = {self.__x}, y={self.__y}, symbol = {self.symbol!r})")

    
class Triangle(Base):
    def __init__(self, x1, y1, x2, y2, x3, y3, symbol: str = "^"):
        self.__x1, self.__y1 = x1 , y1
        self.__x2, self.__y2 = x2 , y2
        self.__x3, self.__y3 = x3 , y3
        self.symbol = symbol

    def get_vertices(self):
        return [(self.__x1, self.__y1), (self.__x2, self.__y2), (self.__x3, self.__y3)]

    def __side_length(self, x_a, y_a, x_b, y_b):
        return math.hypot(x_b - x_a, y_b - y_a)
    
    def get_area(self):
        area = 0.5 * abs(
            self.__x1 * (self.__y2 - self.__y3) + 
            self.__x2 * (self.__y3 - self.__y1) + 
            self.__x3 * (self.__y1 - self.__y2)
        )
        return area

    def point_in_shape(self, px:float, py:float) -> bool:
        def sign(p1_x, p1_y, p2_x, p2_y, p3_x, p3_y):
            return (p1_x - p3_x) *(p2_y - p3_y) - (p2_x - p3_x) * (p1_y - p3_y)
        d1 = sign(px, py, self.__x1, self.__y1, self.__x2, self.__y2)
        d2 = sign(px, py, self.__x2, self.__y2, self.__x3, self.__y3)
        d3 = sign(px, py, self.__x3, self.__y3, self.__x1, self.__y1)
        has_neg = (d1 < 0) or (d2 < 0) or (d3 < 0)
        has_pos = (d1 > 0) or (d2 > 0) or (d3 > 0)
        return not(has_neg and has_pos)

    def draw(self, canvas:"Canvas", char : str = "*") -> None:
        for row in range(canvas.height):
            for col in range(canvas.width):
                if self.point_in_shape(col,row):
                    canvas.set_pixel(row,col,char)

    def __repr__(self) -> str:
        return (f"Triangle(x1={self.__x1}, y1={self.__y1}, x2={self.__x2},"
                f" y2={self.__y2}, x3={self.__x3}, y3={self.__y3},"
                f" symbol={self.symbol!r})")

class CompoundShape(Base):
    def __init__(self, shapes: list[Base] = None):
        self.__shapes: list[Base] = shapes if shapes is not None else []

    def add_shape(self, shape: Base) -> None:
        self.__shapes.append(base)

    def remove_shapes(self, shape: Base) -> None:
        if shape in self.__shapes:
            self.__shapes.remove(shape)

    def get_area(self) -> float:
        return sum(s.get_area() for s in self.__shapes)

    def point_in_shape(self, px:float, py:float) -> bool:
        return any(s.point_in_shape(px,py) for s in self.__shapes)

    def draw(self, canvas: "Canvas") -> None:
        for shape in self.__shapes:
            shape.draw(canvas)

    def __repr__(self) -> str:
        shapes_repr = ",\n   ".join(repr(s) for s in self.shapes)
        return(f"RasterDrawing(width={self.canvas.edith}, height={self.canvas.height})\n# Shapes:\n{shapes_repr}")

class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        # Empty canvas is a matrix with element being the "space" character
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]
    
    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]
            
    def display(self):
        print("\n".join(["".join(row) for row in self.data]))

class RasterDrawing:
    def __init__(self, width: int = 30, height: int=12):
        self.canvas = Canvas(width,height)
        self.shapes: list[Base] = []

    def add_shape(self, shape: Base) -> None:
        self.shapes.append(shape)

    def remove_shape(self, shape: Base) -> None:
        if shape in self.shapes:
            self.shapes.remove(shape)
            
    def paint(self) -> None:
        self.canvas.clear_canvas()
        for shape in self.shapes:
            shape.draw(self.canvas)
        self.canvas.display()

    def __repr__(self) -> str:
        return(f"RasterDrawing.from_repr(width={self.canvas.width}, height={self.canvas.height}, shapes = {self.shapes!r})")

    @classmethod
    def from_repr(cls, width: int, height: int, shapes: list[Base]) -> 'RasterDrawing':
        drawing = cls(width=width, height=height)
        for shape in shapes:
            drawing.add_shape(shape)
        return drawing
```

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b
        
    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"
    
    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()
        
   
def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```
